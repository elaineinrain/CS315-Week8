### Q8: Time frame for a given country

Write a function that given a country name prints out the start and end date of their rankings.

Once you have created the function, call it for a few countries to test it.

```python
def time_frame(country):
    weeks = df[df['country_name'] == country]['week']
    print(f"{country}: {weeks.min()} to {weeks.max()}")

for c in ['Russia', 'South Korea', 'Italy']:
    time_frame(c)
```

### Q9: Is number of weeks equal for each country?

The total number of weeks in the rankings is `len(df['week'].unique())`.

The total number of countries is `len(df['country_name'].unique())`.  

For every week, each country has 20 rows.   

If they all stayed the same number of weeks, then the product of these three weeks would be the same as the total number of rows.  

```python
len(df['week'].unique()) * len(df['country_name'].unique()) * 20 == len(df)
```

We should have expected this to be `False` since we saw in Q8 that Russia stopped being in the rankings after it attacked Ukraine in February 2022.

### New task: rank2score

```python
# create a dictionary with rank as key and score as value. 
# call .get() to get the score corresponding to the rank
def rank2score(rank):
    points = {
        1: 25,
        2: 18,
        3: 15,
        4: 12,
        5: 10,
        6: 8,
        7: 6,
        8: 4,
        9: 2,
        10: 1
    }
    return points.get(rank, 0)
```

```python
df["score"] = df["weekly_rank"].apply(rank2score)
```

```python
# keep only the useful columns using [[ ]]
df = df[["country_name", "week", "weekly_rank", "score", "show_title", "cumulative_weeks_in_top_10"]]
df
```

```python
# create a new df for When Life Gives You Tangerines
df2 = pd.DataFrame()
# calculate total scores grouped by countries
df2 = df.loc[df["show_title"]=="When Life Gives You Tangerines"].groupby("country_name")["score"].sum().reset_index(name="total_points")
# calculate total weeks and map it to the dataframe
weeks = df.loc[df["show_title"]=="When Life Gives You Tangerines"].groupby("country_name")["cumulative_weeks_in_top_10"].sum()
df2["total_weeks"] = df2["country_name"].map(weeks)
# list the rankings
rankingsList = df.loc[df["show_title"]=="When Life Gives You Tangerines"].groupby("country_name")["weekly_rank"].agg(list)
df2["rankings_list"] = df2["country_name"].map(rankingsList)
df2
```
